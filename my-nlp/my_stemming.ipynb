```python
words = ["eaten", "eating", "eats", "easily", "fairly"]
```

PorterStemmer Example

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
for w in words:
    print(f"{w} --> {ps.stem(w)}")
```

RegExpStemmer Example

```python
from nltk.stem import RegexpStemmer
rs = RegexpStemmer('ing$|s$|e$|able$')
for w in words:
    print(f"{w} --> {rs.stem(w)}")
```

SnowBall Stemmer Example

```python
from nltk.stem import SnowballStemmer
ss = SnowballStemmer('english')
for w in words:
    print(f"{w} --> {ss.stem(w)}")
```
