```python
import pandas as pd
```

```python
messages = pd.read_csv('SMSSpamCollection.txt', sep='\t', names=["label", "message"])
messages
```

```python
import nltk
import re
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

Text Preprocessing using Regex, Stopwords and Stemming

```python
corpus =[]
for i in range(len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if word not in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
corpus
```

Using Bag of Words using sklearn CountVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer 
cv = CountVectorizer(max_features=2500) # for Binary BoW use binary=True
```

```python
X = cv.fit_transform(corpus).toarray()
X
```

Using Ngrams

```python
cv.vocabulary_ # top 100 words in the corpus with their indices
```

```python
# Bag of Words model using Ngrams
from sklearn.feature_extraction.text import CountVectorizer 
cv = CountVectorizer(max_features=2500, binary=True, ngram_range=(2,2)) # for Binary BoW use binary=True
X = cv.fit_transform(corpus).toarray()
```

```python
cv.vocabulary_ # top 100 ngrams in the corpus with their indices
```
