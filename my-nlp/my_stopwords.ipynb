```python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords

ps = PorterStemmer()    
```

```python
paragraph = """In computer science, artificial intelligence (AI), sometimes called machine intelligence, is intelligence demonstrated by machines, in contrast to the natural intelligence displayed by humans and animals. Leading AI textbooks define the field as the study of "intelligent agents": any device that perceives its environment and takes actions that maximize its chance of successfully achieving its goals.[1] Colloquially, the term "artificial intelligence" is often used to describe machines (or computers) that mimic "cognitive" functions that humans associate with the human mind, such as "learning" and "problem solving"."""    
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords_list = stopwords.words('english')
stopwords_list
```

```python
corpus = "Hello, world! This is a sample corpus for tokenization. This is a test. Tokenization is fun."
corpus = "Other photographs include two images showing public intellectual Noam Chomsky seated with Epstein on what appears to be an aircraft, and pictures of Bill Gates posing beside a woman whose face is blurred"
sentences = nltk.sent_tokenize(corpus)
```

Perform Stopwords and Porter Stemming

```python
sentences = nltk.sent_tokenize(corpus)

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i]) # Tokenization of words
    words = [ps.stem(word) for word in words if word.lower() not in stopwords_list] # Stemming and Stopword Removal
    sentences[i] = ' '.join(words) # Joining the words back to form the sentence
print(sentences)
```

Using Snowball Stemmer and Stopwords

```python
from nltk.stem import SnowballStemmer
ss = SnowballStemmer('english')

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i]) # Tokenization of words
    words = [ss.stem(word) for word in words if word.lower() not in stopwords_list] # Stemming and Stopword Removal
    sentences[i] = ' '.join(words) # Joining the words back to form the sentence
print(sentences)
```

Using Lemmitization

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i]) # Tokenization of words
    words = [lemmatizer.lemmatize(word,pos='a') for word in words if word.lower() not in stopwords_list] # Lemmatization and Stopword Removal
    sentences[i] = ' '.join(words) # Joining the words back to form the sentence
print(sentences)
```
