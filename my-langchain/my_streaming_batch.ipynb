```python
from dotenv import load_dotenv
import os
import langchain

load_dotenv(override=True)
API_HOST = os.getenv("API_HOST", "github")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite", google_api_key=os.getenv("GOOGLE_API_KEY"))
response = llm.invoke("why do parrots talk")
response.content
```

Streaming

```python
### llm.stream --> returns a generator object which needs to be iterated

for chunk in llm.stream("write me a 200 paragraph on artificial intelligence"):
    print(chunk.text)
```

Batch. Sends inputs to parallel and output comes all at once

```python
response = llm.batch(
    ["why do parrots have colourful feathers",
    "what is the meaning of LSTM",
    "what is quantum computing"
    ],
    config={"max_concurrency":3}
)
for i in response:
    print(i.content)
```
