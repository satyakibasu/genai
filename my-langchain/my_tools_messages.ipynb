```python
from dotenv import load_dotenv
import os
import langchain
from rich import print

load_dotenv(override=True)
API_HOST = os.getenv("API_HOST", "github")

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
        model_name=os.getenv("GITHUB_MODEL", "openai/gpt-4o"),
        openai_api_base="https://models.github.ai/inference",
        openai_api_key=os.environ["GITHUB_TOKEN"],
    )
```

```python
from langchain.tools import tool

@tool ## This binds its to the LLM model as a tool
def get_weather(location:str)->str:
    """This is a weather calling function"""
    return f"The weather is sunny at {location}"


model_with_tool = llm.bind_tools([get_weather])
## You can also use create_agent and pass tools into that function

response = model_with_tool.invoke("what's the weather in Pune")
print(response)

for tool_call in response.tool_calls:
    print(tool_call['name'])
    print(tool_call['args'])

```

```python
from langchain.messages import HumanMessage, SystemMessage, AIMessage

messages = [
    SystemMessage("you are an poet expert"),
    HumanMessage("write a poem on AI")
]

response = llm.invoke(messages)
print(response)
```
