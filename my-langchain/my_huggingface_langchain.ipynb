HuggingFace and Langchain Integration

```python
import os
from dotenv import load_dotenv

load_dotenv()  # take environment variables from .env file
```

Let's call some models on Hugging Face Hub
Access Huggingface models with API 

```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace

repo_id = "HuggingFaceH4/zephyr-7b-beta"
llm = HuggingFaceEndpoint(
    repo_id=repo_id,
    huggingfacehub_api_token=os.getenv("HF_TOKEN"),
    temperature=0.7,
    max_new_tokens=512,
)
llm = ChatHuggingFace(llm=llm)
```

```python
llm.invoke("what is machine learning?")
```

```python
from langchain_core.prompts import PromptTemplate

template = """You are a helpful assistant that helps people find information.
{question}"""

prompt = PromptTemplate(
    input_variables=["question"],
    template=template,
)
```

```python
chain = prompt | llm
response = chain.invoke({"question": "Explain the theory of relativity."})
print(response)
```
