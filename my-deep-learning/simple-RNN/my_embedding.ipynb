```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
# sentences
sentences = [
    'I love machine learning',
    'Deep learning is a subset of machine learning',
    'Natural language processing is fun'
]
```

One Hot Representation

```python
vocabulary_size = 10000
one_hot_encoded_repr = [one_hot(sentence, vocabulary_size) for sentence in sentences]
one_hot_encoded_repr
```

Word Embedding Represenatition

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
## Let's pad the sequences to ensure uniform input length
max_length = 10
padded_sequences = pad_sequences(one_hot_encoded_repr, maxlen=max_length, padding='post')
padded_sequences
```

```python
embedding_dim = 10 # Feature representation using Embedding layer

model = Sequential()
model.add(Embedding(input_dim=vocabulary_size, output_dim=embedding_dim, input_length=max_length))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
model.predict(padded_sequences)
```
